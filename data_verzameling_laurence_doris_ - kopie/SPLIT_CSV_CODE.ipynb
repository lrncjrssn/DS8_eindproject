```python
metadata = []

for i in ['Asfalt', 'Brug', 'Gras', 'Stenen']:
    for j in [0, 0.5, 1, 1.5, 2]:
        filepath = f"{i} {j}\Raw Data.csv"
        df = pd.read_csv(filepath)
        #map_output = filepath + "_split" ## andere naam
        #os.makedirs(map_output, exist_ok=True)
        df['second'] = df['Time (s)'].astype(int)
        for second, group in df.groupby('second'):
            output_file = f"{i}_{j}_second_{second}.csv"
            group.drop(columns=['second'], inplace=True)
            group.to_csv(output_file, index=False)
            print(f"Saved {second}.csv with {len(group)} rows.")
            
            metadata.append({
            "cut_file": f"{i}_{j}_second_{second}.csv",
            "road_types": i,
            "backpack_weight": j,
            'weather (degrees)': '11'
        })
            ## weart
```

```python
metadata = pd.DataFrame(metadata)
metadata.head()
```
