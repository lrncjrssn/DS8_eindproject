```python
import pandas as pd
import glob
import os

pad = r"C:\Users\j4985\OneDrive\Python\DS8\1sec_cuts/*.csv"
bestanden = sorted(glob.glob(pad))  

bestanden = bestanden[1:]

print(len(bestanden))

rows = []

for f in bestanden:
    df = pd.read_csv(f, nrows=1)
    df["source_file"] = os.path.basename(f)
    rows.append(df)

result = pd.concat(rows, ignore_index=True)
result.to_csv("eerste_regels_zonder_eerste.csv", index=False)
```
